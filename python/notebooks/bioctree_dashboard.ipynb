## 3. Data Loading and Preprocessing Functions

Robust data management system with validation, error handling, and preprocessing capabilities.

````python
import networkx as nx
import numpy as np
import matplotlib.pyplot as plt
from dataclasses import dataclass
from typing import Dict, Any, Optional, Tuple

# Check for widget availability
try:
    from ipywidgets import interact_manual
    HAS_WIDGETS = True
except ImportError:
    HAS_WIDGETS = False
    print("⚠️ IPyWidgets not available - using manual interface")

@dataclass
class GraphData:
    """Container for graph and signal data"""
    graph: nx.Graph
    signals: Dict[str, np.ndarray]
    positions: Dict[int, Tuple[float, float]]
    metadata: Dict[str, Any]

class DataManager:
    """Manages graph and signal generation for the dashboard"""
    
    def __init__(self):
        self.current_data: Optional[GraphData] = None
        self.graph_generators = {
            'Stanford Bunny': self._generate_bunny_graph,
            'Random Graph': self._generate_random_graph,
            'Grid Graph': self._generate_grid_graph,
            'Ring Graph': self._generate_ring_graph
        }
        self.signal_generators = {
            'Smooth': self._generate_smooth_signal,
            'Oscillatory': self._generate_oscillatory_signal,
            'Random': self._generate_random_signal,
            'Localized': self._generate_localized_signal
        }
    
    def _generate_bunny_graph(self, n_vertices: int) -> Tuple[nx.Graph, Dict]:
        """Generate Stanford Bunny-like graph"""
        # Create a more complex geometric graph that resembles biological data
        G = nx.random_geometric_graph(n_vertices, radius=0.3, seed=42)
        
        # Add some long-range connections for brain-like connectivity
        for i in range(n_vertices // 10):
            u, v = np.random.choice(n_vertices, 2, replace=False)
            if not G.has_edge(u, v):
                G.add_edge(u, v)
        
        # Create 3D-like positions projected to 2D
        pos = {}
        theta = np.linspace(0, 2*np.pi, n_vertices)
        phi = np.linspace(0, np.pi, n_vertices)
        
        for i, node in enumerate(G.nodes()):
            # Create bunny-like shape
            t = theta[i]
            p = phi[i]
            x = 0.5 * (2 + np.cos(t)) * np.sin(p)
            y = 0.5 * (2 + np.cos(t)) * np.cos(p)
            pos[node] = (x, y)
        
        return G, pos
    
    def _generate_random_graph(self, n_vertices: int) -> Tuple[nx.Graph, Dict]:
        """Generate random graph"""
        G = nx.erdos_renyi_graph(n_vertices, 0.1, seed=42)
        pos = nx.spring_layout(G, seed=42)
        return G, pos
    
    def _generate_grid_graph(self, n_vertices: int) -> Tuple[nx.Graph, Dict]:
        """Generate grid graph"""
        n = int(np.sqrt(n_vertices))
        G = nx.grid_2d_graph(n, n)
        # Convert to simple graph with integer nodes
        G = nx.convert_node_labels_to_integers(G)
        pos = nx.spring_layout(G, seed=42)
        return G, pos
    
    def _generate_ring_graph(self, n_vertices: int) -> Tuple[nx.Graph, Dict]:
        """Generate ring graph"""
        G = nx.cycle_graph(n_vertices)
        pos = nx.circular_layout(G)
        return G, pos
    
    def _generate_smooth_signal(self, graph: nx.Graph, noise_level: float) -> np.ndarray:
        """Generate smooth signal using graph structure"""
        n = len(graph.nodes())
        
        # Create smooth signal using graph Laplacian eigenvectors
        try:
            L = nx.normalized_laplacian_matrix(graph).toarray()
            eigenvals, eigenvecs = np.linalg.eigh(L)
            
            # Use low-frequency components for smooth signal
            signal = eigenvecs[:, 1] + 0.3 * eigenvecs[:, 2]
            signal = signal / np.max(np.abs(signal))
        except:
            # Fallback: use spatial smoothness
            signal = np.random.randn(n)
            # Simple smoothing
            for _ in range(5):
                new_signal = signal.copy()
                for node in graph.nodes():
                    neighbors = list(graph.neighbors(node))
                    if neighbors:
                        new_signal[node] = 0.7 * signal[node] + 0.3 * np.mean([signal[nb] for nb in neighbors])
                signal = new_signal
        
        # Add noise
        noise = noise_level * np.random.randn(n)
        return signal + noise
    
    def _generate_oscillatory_signal(self, graph: nx.Graph, noise_level: float) -> np.ndarray:
        """Generate oscillatory signal"""
        n = len(graph.nodes())
        
        try:
            L = nx.normalized_laplacian_matrix(graph).toarray()
            eigenvals, eigenvecs = np.linalg.eigh(L)
            
            # Use mid-frequency components
            signal = eigenvecs[:, n//4] + 0.5 * eigenvecs[:, n//3]
            signal = signal / np.max(np.abs(signal))
        except:
            # Fallback: sinusoidal pattern
            signal = np.sin(np.linspace(0, 4*np.pi, n)) + 0.5 * np.cos(np.linspace(0, 6*np.pi, n))
        
        noise = noise_level * np.random.randn(n)
        return signal + noise
    
    def _generate_random_signal(self, graph: nx.Graph, noise_level: float) -> np.ndarray:
        """Generate random signal"""
        n = len(graph.nodes())
        signal = np.random.randn(n)
        noise = noise_level * np.random.randn(n)
        return signal + noise
    
    def _generate_localized_signal(self, graph: nx.Graph, noise_level: float) -> np.ndarray:
        """Generate localized signal"""
        n = len(graph.nodes())
        signal = np.zeros(n)
        
        # Create a few localized peaks
        n_peaks = max(1, n // 20)
        peak_nodes = np.random.choice(n, n_peaks, replace=False)
        
        for peak_node in peak_nodes:
            # Add signal at peak and propagate to neighbors
            signal[peak_node] = 1.0
            for neighbor in graph.neighbors(peak_node):
                signal[neighbor] += 0.5
                for nn in graph.neighbors(neighbor):
                    signal[nn] += 0.2
        
        # Normalize
        signal = signal / np.max(np.abs(signal) + 1e-10)
        
        noise = noise_level * np.random.randn(n)
        return signal + noise
    
    def generate_data(self, graph_type: str, n_vertices: int, signal_type: str, noise_level: float) -> GraphData:
        """Generate graph and signal data"""
        # Generate graph
        if graph_type in self.graph_generators:
            graph, pos = self.graph_generators[graph_type](n_vertices)
        else:
            graph, pos = self._generate_random_graph(n_vertices)
        
        # Generate signals
        signals = {}
        if signal_type in self.signal_generators:
            signals[signal_type] = self.signal_generators[signal_type](graph, noise_level)
        else:
            signals['Random'] = self._generate_random_signal(graph, noise_level)
        
        # Store metadata
        metadata = {
            'graph_type': graph_type,
            'n_vertices': n_vertices,
            'signal_type': signal_type,
            'noise_level': noise_level,
            'n_edges': graph.number_of_edges(),
            'avg_degree': np.mean([graph.degree(node) for node in graph.nodes()])
        }
        
        self.current_data = GraphData(
            graph=graph,
            signals=signals,
            positions=pos,
            metadata=metadata
        )
        
        return self.current_data
    
    def get_current_data(self) -> Optional[GraphData]:
        """Get current data"""
        return self.current_data
    
    def has_data(self) -> bool:
        """Check if data is available"""
        return self.current_data is not None

# Create global data manager
data_manager = DataManager()

def generate_data_manual(graph_type='Stanford Bunny', n_vertices=50, signal_type='Smooth', noise_level=0.1):
    """Manual data generation function"""
    print(f"� Generating {graph_type} with {n_vertices} vertices...")
    print(f"📊 Signal type: {signal_type}, Noise level: {noise_level}")
    
    data = data_manager.generate_data(graph_type, n_vertices, signal_type, noise_level)
    
    print(f"✅ Generated graph with {data.graph.number_of_nodes()} nodes and {data.graph.number_of_edges()} edges")
    print(f"📈 Average degree: {data.metadata['avg_degree']:.2f}")
    
    return data

# Interactive interface (only if widgets available)
if HAS_WIDGETS:
    try:
        @interact_manual(
            graph_type=['Stanford Bunny', 'Random Graph', 'Grid Graph', 'Ring Graph'],
            n_vertices=(10, 200, 10),
            signal_type=['Smooth', 'Oscillatory', 'Random', 'Localized'],
            noise_level=(0.0, 0.5, 0.05)
        )
        def data_interface(graph_type='Stanford Bunny', n_vertices=50, signal_type='Smooth', noise_level=0.1):
            """Interactive data generation interface"""
            return generate_data_manual(graph_type, n_vertices, signal_type, noise_level)
        
        print("🎛️ Interactive data generation interface ready!")
        print("📝 Use the controls above to generate graph data")
    
    except Exception as e:
        print(f"⚠️ Widget interface failed: {e}")
        print("🔧 Using manual interface instead")
        HAS_WIDGETS = False

if not HAS_WIDGETS:
    print("� Manual Interface Available:")
    print("   generate_data_manual(graph_type, n_vertices, signal_type, noise_level)")
    print("   Example: generate_data_manual('Stanford Bunny', 50, 'Smooth', 0.1)")
    print()
    print("🚀 Generating demo data...")
    demo_data = generate_data_manual('Stanford Bunny', 50, 'Smooth', 0.1)

print(f"✅ DataManager ready! (Widgets: {'✅' if HAS_WIDGETS else '❌'})")
````

## 5. Graph Signal Processing Core Functions

Implementation of core GSP operations with comprehensive analysis and filtering capabilities.

````python
class GraphSignalProcessor:
    """Comprehensive graph signal processing operations"""
    
    def __init__(self):
        self.processing_history = []
        self.filters_cache = {}
        
    def compute_graph_fourier_transform(self, graph, signal):
        """Compute the Graph Fourier Transform"""
        
        # Compute eigendecomposition of Laplacian
        eigenvals, eigenvecs = linalg.eigh(graph.L)
        
        # Compute GFT coefficients
        gft_coeffs = eigenvecs.T @ signal
        
        # Store for analysis
        gft_data = {
            'eigenvals': eigenvals,
            'eigenvecs': eigenvecs,
            'coefficients': gft_coeffs,
            'energy': np.abs(gft_coeffs)**2,
            'signal_energy': np.sum(np.abs(gft_coeffs)**2)
        }
        
        return gft_data
    
    def apply_graph_filter(self, graph, signal, filter_type='low-pass', 
                          cutoff=0.3, **kwargs):
        """Apply various graph filters to signal"""
        
        # Get eigendecomposition
        eigenvals, eigenvecs = linalg.eigh(graph.L)
        gft_coeffs = eigenvecs.T @ signal
        
        # Normalize eigenvalues to [0,1]
        eigenvals_norm = eigenvals / np.max(eigenvals) if np.max(eigenvals) > 0 else eigenvals
        
        # Define filter response
        if filter_type == 'low-pass':
            # Ideal low-pass filter
            filter_response = (eigenvals_norm <= cutoff).astype(float)
            
        elif filter_type == 'high-pass':
            # Ideal high-pass filter
            filter_response = (eigenvals_norm >= cutoff).astype(float)
            
        elif filter_type == 'band-pass':
            # Band-pass filter
            low_cutoff = kwargs.get('low_cutoff', 0.2)
            high_cutoff = kwargs.get('high_cutoff', 0.8)
            filter_response = ((eigenvals_norm >= low_cutoff) & 
                             (eigenvals_norm <= high_cutoff)).astype(float)
            
        elif filter_type == 'heat-kernel':
            # Heat kernel filter
            tau = kwargs.get('tau', 10)
            filter_response = np.exp(-tau * eigenvals_norm)
            
        else:
            # Default: no filtering
            filter_response = np.ones_like(eigenvals_norm)
        
        # Apply filter in frequency domain
        filtered_coeffs = gft_coeffs * filter_response
        
        # Inverse GFT
        filtered_signal = eigenvecs @ filtered_coeffs
        
        # Store processing information
        filter_info = {
            'filter_type': filter_type,
            'cutoff': cutoff,
            'filter_response': filter_response,
            'eigenvals': eigenvals_norm,
            'original_energy': np.sum(np.abs(gft_coeffs)**2),
            'filtered_energy': np.sum(np.abs(filtered_coeffs)**2)
        }
        
        return filtered_signal, filter_info
    
    def compute_graph_gradient(self, graph, signal):
        """Compute graph gradient (spatial derivative)"""
        
        if hasattr(graph, 'W'):
            W = graph.W
        else:
            W = graph.adjacency
            
        if sparse.issparse(W):
            W = W.toarray()
        
        # Compute gradient magnitude for each edge
        gradients = []
        edge_list = []
        
        for i in range(graph.N):
            for j in range(i+1, graph.N):
                if W[i, j] > 0:  # Edge exists
                    grad_val = abs(signal[i] - signal[j]) * np.sqrt(W[i, j])
                    gradients.append(grad_val)
                    edge_list.append((i, j))
        
        total_variation = np.sum(gradients)
        
        return {
            'total_variation': total_variation,
            'edge_gradients': gradients,
            'edge_list': edge_list,
            'mean_gradient': np.mean(gradients) if gradients else 0,
            'max_gradient': np.max(gradients) if gradients else 0
        }
    
    def detect_signal_patterns(self, graph, signal):
        """Detect various patterns in graph signals"""
        
        # Compute spectral analysis
        eigenvals, eigenvecs = linalg.eigh(graph.L)
        gft_coeffs = eigenvecs.T @ signal
        
        # Energy distribution analysis
        energy = np.abs(gft_coeffs)**2
        total_energy = np.sum(energy)
        
        # Low/high frequency energy
        mid_point = len(eigenvals) // 2
        low_freq_energy = np.sum(energy[:mid_point]) / total_energy
        high_freq_energy = np.sum(energy[mid_point:]) / total_energy
        
        # Detect patterns
        patterns = {
            'is_smooth': low_freq_energy > 0.8,
            'is_oscillatory': high_freq_energy > 0.6,
            'is_localized': np.max(energy) > 0.5 * total_energy,
            'spectral_concentration': np.sum(energy > 0.1 * np.max(energy)) / len(energy),
            'low_freq_energy': low_freq_energy,
            'high_freq_energy': high_freq_energy
        }
        
        # Spatial patterns
        signal_stats = {
            'mean': np.mean(signal),
            'std': np.std(signal),
            'skewness': scipy.stats.skew(signal) if 'scipy.stats' in globals() else 0,
            'kurtosis': scipy.stats.kurtosis(signal) if 'scipy.stats' in globals() else 0,
            'range': np.max(signal) - np.min(signal)
        }
        
        return {
            'spectral_patterns': patterns,
            'spatial_stats': signal_stats,
            'energy_distribution': energy / total_energy
        }
    
    def run_comprehensive_analysis(self, graph, signal):
        """Run complete analysis pipeline"""
        
        analysis_results = {}
        
        # Basic statistics
        analysis_results['basic_stats'] = {
            'signal_length': len(signal),
            'graph_size': graph.N,
            'signal_mean': np.mean(signal),
            'signal_std': np.std(signal),
            'signal_range': np.max(signal) - np.min(signal)
        }
        
        # Spectral analysis
        analysis_results['spectral'] = self.compute_graph_fourier_transform(graph, signal)
        
        # Gradient analysis
        analysis_results['gradient'] = self.compute_graph_gradient(graph, signal)
        
        # Pattern detection
        analysis_results['patterns'] = self.detect_signal_patterns(graph, signal)
        
        # Filter demonstrations
        filter_types = ['low-pass', 'high-pass', 'heat-kernel']
        analysis_results['filters'] = {}
        
        for ftype in filter_types:
            try:
                filtered_sig, filter_info = self.apply_graph_filter(graph, signal, ftype)
                analysis_results['filters'][ftype] = {
                    'filtered_signal': filtered_sig,
                    'filter_info': filter_info
                }
            except Exception as e:
                analysis_results['filters'][ftype] = {'error': str(e)}
        
        # Store in history
        self.processing_history.append({
            'timestamp': time.strftime('%Y-%m-%d %H:%M:%S'),
            'analysis': analysis_results
        })
        
        return analysis_results

# Initialize signal processor
gsp_processor = GraphSignalProcessor()

print("🔧 Graph Signal Processor ready!")
````

## 7. Function Development and Testing Framework

Comprehensive testing environment for developing and validating new graph signal processing functions.

````python
class DevelopmentFramework:
    """Testing and development framework for new GSP functions"""
    
    def __init__(self):
        self.test_results = []
        self.benchmark_data = {}
        
    def create_test_suite(self):
        """Create comprehensive test suite for GSP functions"""
        
        class GSPTestSuite(unittest.TestCase):
            
            def setUp(self):
                """Set up test fixtures"""
                # Create test graph
                self.test_graph = stanford_bunny() 
                # Create test signals
                self.smooth_signal = np.random.randn(self.test_graph.N)
                self.zero_signal = np.zeros(self.test_graph.N)
                self.constant_signal = np.ones(self.test_graph.N)
                
            def test_graph_properties(self):
                """Test basic graph properties"""
                self.assertGreater(self.test_graph.N, 0)
                self.assertIsNotNone(self.test_graph.L)
                self.assertEqual(self.test_graph.L.shape[0], self.test_graph.N)
                
            def test_signal_processing(self):
                """Test signal processing functions"""
                # Test gradient computation
                grad_result = gsp_processor.compute_graph_gradient(self.test_graph, self.smooth_signal)
                self.assertIn('total_variation', grad_result)
                self.assertGreaterEqual(grad_result['total_variation'], 0)
                
            def test_filtering(self):
                """Test graph filtering"""
                filtered_signal, _ = gsp_processor.apply_graph_filter(
                    self.test_graph, self.smooth_signal, 'low-pass', 0.3)
                self.assertEqual(len(filtered_signal), self.test_graph.N)
                
            def test_fourier_transform(self):
                """Test Graph Fourier Transform"""
                gft_data = gsp_processor.compute_graph_fourier_transform(
                    self.test_graph, self.smooth_signal)
                self.assertIn('coefficients', gft_data)
                self.assertEqual(len(gft_data['coefficients']), self.test_graph.N)
        
        return GSPTestSuite
    
    def run_tests(self, verbose=True):
        """Run the complete test suite"""
        
        print("🧪 Running GSP function tests...")
        
        # Create test suite
        TestSuite = self.create_test_suite()
        
        # Run tests
        loader = unittest.TestLoader()
        suite = loader.loadTestsFromTestCase(TestSuite)
        
        # Capture test results
        import io
        test_output = io.StringIO()
        runner = unittest.TextTestRunner(stream=test_output, verbosity=2)
        result = runner.run(suite)
        
        # Store results
        test_summary = {
            'timestamp': time.strftime('%Y-%m-%d %H:%M:%S'),
            'tests_run': result.testsRun,
            'failures': len(result.failures),
            'errors': len(result.errors),
            'success_rate': (result.testsRun - len(result.failures) - len(result.errors)) / result.testsRun if result.testsRun > 0 else 0,
            'output': test_output.getvalue()
        }
        
        self.test_results.append(test_summary)
        
        if verbose:
            print(f"✅ Tests completed: {test_summary['tests_run']} run")
            print(f"📊 Success rate: {test_summary['success_rate']:.1%}")
            if test_summary['failures'] > 0:
                print(f"❌ Failures: {test_summary['failures']}")
            if test_summary['errors'] > 0:
                print(f"🚫 Errors: {test_summary['errors']}")
                
        return test_summary
    
    def benchmark_function(self, func, *args, n_runs=10, **kwargs):
        """Benchmark a function's performance"""
        
        print(f"⏱️ Benchmarking function: {func.__name__}")
        
        times = []
        for i in range(n_runs):
            start_time = time.time()
            try:
                result = func(*args, **kwargs)
                end_time = time.time()
                times.append(end_time - start_time)
            except Exception as e:
                print(f"❌ Error in run {i+1}: {e}")
                return None
        
        benchmark_stats = {
            'function_name': func.__name__,
            'n_runs': n_runs,
            'mean_time': np.mean(times),
            'std_time': np.std(times),
            'min_time': np.min(times),
            'max_time': np.max(times),
            'total_time': np.sum(times)
        }
        
        self.benchmark_data[func.__name__] = benchmark_stats
        
        print(f"📈 Average time: {benchmark_stats['mean_time']:.4f}s ± {benchmark_stats['std_time']:.4f}s")
        print(f"📈 Range: [{benchmark_stats['min_time']:.4f}s, {benchmark_stats['max_time']:.4f}s]")
        
        return benchmark_stats
    
    def validate_new_function(self, func, test_graph=None, test_signal=None):
        """Validate a new GSP function"""
        
        print(f"🔍 Validating function: {func.__name__}")
        
        # Use default test data if not provided
        if test_graph is None:
            test_graph = stanford_bunny()
        if test_signal is None:
            test_signal = np.random.randn(test_graph.N)
        
        validation_results = {
            'function_name': func.__name__,
            'timestamp': time.strftime('%Y-%m-%d %H:%M:%S'),
            'tests': {}
        }
        
        # Test 1: Basic execution
        try:
            result = func(test_graph, test_signal)
            validation_results['tests']['basic_execution'] = {'status': 'PASS', 'result_type': type(result).__name__}
        except Exception as e:
            validation_results['tests']['basic_execution'] = {'status': 'FAIL', 'error': str(e)}
            return validation_results
        
        # Test 2: Output validity
        try:
            if isinstance(result, np.ndarray):
                if len(result) == test_graph.N:
                    validation_results['tests']['output_size'] = {'status': 'PASS'}
                else:
                    validation_results['tests']['output_size'] = {'status': 'FAIL', 'expected': test_graph.N, 'got': len(result)}
            else:
                validation_results['tests']['output_size'] = {'status': 'SKIP', 'reason': 'Non-array output'}
        except Exception as e:
            validation_results['tests']['output_size'] = {'status': 'ERROR', 'error': str(e)}
        
        # Test 3: Numerical stability
        try:
            if isinstance(result, np.ndarray):
                if np.all(np.isfinite(result)):
                    validation_results['tests']['numerical_stability'] = {'status': 'PASS'}
                else:
                    validation_results['tests']['numerical_stability'] = {'status': 'FAIL', 'reason': 'Contains NaN or Inf'}
            else:
                validation_results['tests']['numerical_stability'] = {'status': 'SKIP', 'reason': 'Non-array output'}
        except Exception as e:
            validation_results['tests']['numerical_stability'] = {'status': 'ERROR', 'error': str(e)}
        
        # Test 4: Performance benchmark
        try:
            benchmark_stats = self.benchmark_function(func, test_graph, test_signal, n_runs=5)
            if benchmark_stats['mean_time'] < 1.0:  # Reasonable performance threshold
                validation_results['tests']['performance'] = {'status': 'PASS', 'time': benchmark_stats['mean_time']}
            else:
                validation_results['tests']['performance'] = {'status': 'WARN', 'time': benchmark_stats['mean_time'], 'threshold': 1.0}
        except Exception as e:
            validation_results['tests']['performance'] = {'status': 'ERROR', 'error': str(e)}
        
        # Summary
        passed_tests = sum(1 for test in validation_results['tests'].values() if test['status'] == 'PASS')
        total_tests = len(validation_results['tests'])
        
        print(f"✅ Validation complete: {passed_tests}/{total_tests} tests passed")
        
        return validation_results
    
    def create_development_template(self, function_name):
        """Create a template for new function development"""
        
        template = f'''
def {function_name}(graph, signal, **kwargs):
    """
    {function_name.replace('_', ' ').title()} - Graph Signal Processing Function
    
    Parameters:
    -----------
    graph : Graph object
        Input graph with properties N (number of vertices), W (adjacency), L (Laplacian)
    signal : numpy.ndarray
        Input signal of length graph.N
    **kwargs : dict
        Additional parameters
    
    Returns:
    --------
    result : numpy.ndarray or dict
        Processed result
    
    Examples:
    ---------
    >>> graph = stanford_bunny()
    >>> signal = np.random.randn(graph.N)
    >>> result = {function_name}(graph, signal)
    """
    
    # Input validation
    if not hasattr(graph, 'N') or not hasattr(graph, 'L'):
        raise ValueError("Invalid graph structure")
    
    if len(signal) != graph.N:
        raise ValueError(f"Signal length ({{len(signal)}}) must match graph size ({{graph.N}})")
    
    # TODO: Implement your algorithm here
    
    # Example: Identity operation
    result = signal.copy()
    
    return result

# Test the new function
if __name__ == "__main__":
    # Create test data
    test_graph = stanford_bunny()
    test_signal = np.random.randn(test_graph.N)
    
    # Run function
    result = {function_name}(test_graph, test_signal)
    print(f"Function {{'{function_name}'}} completed successfully")
    print(f"Input shape: {{test_signal.shape}}, Output shape: {{result.shape if hasattr(result, 'shape') else type(result)}}")
        '''
        
        return template
    
    def get_development_report(self):
        """Generate comprehensive development report"""
        
        report = f"""
        📋 **Development Framework Report**
        Generated: {time.strftime('%Y-%m-%d %H:%M:%S')}
        
        🧪 **Test Results Summary**
        - Total test sessions: {len(self.test_results)}
        """
        
        if self.test_results:
            latest_test = self.test_results[-1]
            report += f"""
        - Latest test success rate: {latest_test['success_rate']:.1%}
        - Tests run: {latest_test['tests_run']}
        - Failures: {latest_test['failures']}
        - Errors: {latest_test['errors']}
            """
        
        report += f"""
        
        ⏱️ **Benchmark Results**
        - Functions benchmarked: {len(self.benchmark_data)}
        """
        
        for func_name, stats in self.benchmark_data.items():
            report += f"""
        - {func_name}: {stats['mean_time']:.4f}s ± {stats['std_time']:.4f}s
            """
        
        return report

# Initialize development framework
dev_framework = DevelopmentFramework()

# Interactive testing interface
@interact_manual(
    run_tests=False,
    benchmark_gsp=False,
    create_template=False,
    template_name='new_gsp_function'
)
def development_interface(run_tests=False, benchmark_gsp=False, create_template=False, template_name='new_gsp_function'):
    """Interactive development interface"""
    
    if run_tests:
        dev_framework.run_tests()
    
    if benchmark_gsp and data_manager.current_graph is not None:
        print("\\n⏱️ Running benchmarks...")
        dev_framework.benchmark_function(
            gsp_processor.compute_graph_fourier_transform,
            data_manager.current_graph, data_manager.current_signal
        )
        
        dev_framework.benchmark_function(
            gsp_processor.compute_graph_gradient,
            data_manager.current_graph, data_manager.current_signal
        )
    
    if create_template:
        template = dev_framework.create_development_template(template_name)
        print(f"\\n📝 Function template created for '{template_name}':")
        print("\\n" + "="*60)
        print(template)
        print("="*60)

print("🔬 Development framework ready!")
````

````python
class ExportManager:
    """Comprehensive export and reporting system"""
    
    def __init__(self):
        self.export_history = []
        self.report_templates = self._setup_report_templates()
        
        # Create exports directory
        self.export_dir = Path("../exports")
        self.export_dir.mkdir(exist_ok=True)
        
    def _setup_report_templates(self):
        """Setup report templates"""
        return {
            'analysis_report': """
# Graph Signal Processing Analysis Report

**Generated**: {timestamp}
**Graph Type**: {graph_type}
**Signal Type**: {signal_type}

## Executive Summary
{executive_summary}

## Graph Properties
- **Vertices**: {n_vertices}
- **Edges**: {n_edges}
- **Graph Type**: {graph_type}
- **Connectivity**: {connectivity}

## Signal Analysis
- **Signal Type**: {signal_type}
- **Noise Level**: {noise_level}
- **Total Variation**: {total_variation:.4f}
- **Signal Energy**: {signal_energy:.4f}

## Spectral Analysis
- **Low Frequency Energy**: {low_freq_energy:.2%}
- **High Frequency Energy**: {high_freq_energy:.2%}
- **Spectral Bandwidth**: {spectral_bandwidth}

## Key Findings
{key_findings}

## Recommendations
{recommendations}

---
*Report generated by Bioctree Dashboard v1.0*
""",
            'comparison_report': """
# Comparative Analysis Report

**Generated**: {timestamp}

## Comparison Overview
{comparison_overview}

## Methods Compared
{methods_compared}

## Performance Metrics
{performance_metrics}

## Results Summary
{results_summary}

## Conclusions
{conclusions}
"""
        }
    
    def export_figure(self, fig, filename, formats=['png', 'pdf'], dpi=300, **kwargs):
        """Export matplotlib figure in multiple formats"""
        
        exported_files = []
        
        for fmt in formats:
            filepath = self.export_dir / f"{filename}.{fmt}"
            try:
                if fmt == 'pdf':
                    fig.savefig(filepath, format='pdf', bbox_inches='tight', dpi=dpi, **kwargs)
                else:
                    fig.savefig(filepath, format=fmt, dpi=dpi, bbox_inches='tight', **kwargs)
                
                exported_files.append(str(filepath))
                print(f"✅ Exported: {filepath}")
                
            except Exception as e:
                print(f"❌ Error exporting {fmt}: {e}")
        
        return exported_files
    
    def export_plotly_figure(self, fig, filename, formats=['html', 'png']):
        """Export Plotly figure in multiple formats"""
        
        exported_files = []
        
        for fmt in formats:
            filepath = self.export_dir / f"{filename}.{fmt}"
            try:
                if fmt == 'html':
                    fig.write_html(str(filepath))
                elif fmt == 'png':
                    fig.write_image(str(filepath))
                elif fmt == 'pdf':
                    fig.write_image(str(filepath))
                
                exported_files.append(str(filepath))
                print(f"✅ Exported: {filepath}")
                
            except Exception as e:
                print(f"❌ Error exporting {fmt}: {e}")
        
        return exported_files
    
    def export_data(self, data, filename, formats=['csv', 'json', 'pickle']):
        """Export data in multiple formats"""
        
        exported_files = []
        
        for fmt in formats:
            filepath = self.export_dir / f"{filename}.{fmt}"
            try:
                if fmt == 'csv' and isinstance(data, (dict, list)):
                    df = pd.DataFrame(data)
                    df.to_csv(filepath, index=False)
                elif fmt == 'json':
                    import json
                    with open(filepath, 'w') as f:
                        json.dump(data, f, indent=2, default=str)
                elif fmt == 'pickle':
                    import pickle
                    with open(filepath, 'wb') as f:
                        pickle.dump(data, f)
                elif fmt == 'csv' and hasattr(data, 'to_csv'):
                    data.to_csv(filepath, index=False)
                
                exported_files.append(str(filepath))
                print(f"✅ Exported: {filepath}")
                
            except Exception as e:
                print(f"❌ Error exporting {fmt}: {e}")
        
        return exported_files
    
    def generate_analysis_report(self, analysis_results, graph_metadata, signal_metadata):
        """Generate comprehensive analysis report"""
        
        # Extract key information
        graph_type = graph_metadata.get('graph_type', 'Unknown')
        signal_type = signal_metadata.get('signal_type', 'Unknown')
        noise_level = signal_metadata.get('noise_level', 0)
        
        basic_stats = analysis_results.get('basic_stats', {})
        spectral = analysis_results.get('spectral', {})
        gradient = analysis_results.get('gradient', {})
        patterns = analysis_results.get('patterns', {}).get('spectral_patterns', {})
        
        # Generate executive summary
        if patterns.get('is_smooth', False):
            signal_char = "smooth, dominated by low-frequency components"
        elif patterns.get('is_oscillatory', False):
            signal_char = "oscillatory, with significant high-frequency content"
        else:
            signal_char = "mixed frequency content"
        
        executive_summary = f"""
This analysis examines a {signal_type.lower()} signal on a {graph_type.lower()} graph with {basic_stats.get('graph_size', 0)} vertices.
The signal exhibits {signal_char} with a total variation of {gradient.get('total_variation', 0):.4f}.
        """.strip()
        
        # Generate key findings
        key_findings = f"""
1. **Signal Characteristics**: The signal is primarily {signal_char}
2. **Spatial Variation**: Total variation of {gradient.get('total_variation', 0):.4f} indicates {'high' if gradient.get('total_variation', 0) > 1 else 'moderate'} spatial variation
3. **Spectral Distribution**: {patterns.get('low_freq_energy', 0):.1%} energy in low frequencies, {patterns.get('high_freq_energy', 0):.1%} in high frequencies
4. **Signal Quality**: Noise level of {noise_level} {'may affect' if noise_level > 0.3 else 'minimally impacts'} analysis results
        """.strip()
        
        # Generate recommendations
        recommendations = []
        if patterns.get('low_freq_energy', 0) > 0.8:
            recommendations.append("Consider low-pass filtering to reduce noise while preserving signal structure")
        if gradient.get('total_variation', 0) > 2:
            recommendations.append("High spatial variation detected - consider smoothing or regularization")
        if noise_level > 0.3:
            recommendations.append("High noise level - apply denoising techniques before analysis")
        
        recommendations_text = "\\n".join([f"{i+1}. {rec}" for i, rec in enumerate(recommendations)])
        
        # Fill report template
        report = self.report_templates['analysis_report'].format(
            timestamp=time.strftime('%Y-%m-%d %H:%M:%S'),
            graph_type=graph_type,
            signal_type=signal_type,
            executive_summary=executive_summary,
            n_vertices=basic_stats.get('graph_size', 0),
            n_edges=len(gradient.get('edge_list', [])),
            connectivity='Connected' if gradient.get('total_variation', 0) > 0 else 'Disconnected',
            noise_level=noise_level,
            total_variation=gradient.get('total_variation', 0),
            signal_energy=spectral.get('signal_energy', 0),
            low_freq_energy=patterns.get('low_freq_energy', 0),
            high_freq_energy=patterns.get('high_freq_energy', 0),
            spectral_bandwidth=len(spectral.get('eigenvals', [])),
            key_findings=key_findings,
            recommendations=recommendations_text if recommendations else "No specific recommendations at this time."
        )
        
        return report
    
    def create_presentation_slides(self, analysis_results, figures=None):
        """Create presentation slides from analysis"""
        
        slides = f"""
# Graph Signal Processing Analysis
## Bioctree Dashboard Results

---

## Overview
- **Analysis Date**: {time.strftime('%Y-%m-%d')}
- **Graph**: {data_manager.metadata.get('graph_type', 'Unknown')} ({data_manager.current_graph.N if data_manager.current_graph else 0} vertices)
- **Signal**: {data_manager.metadata.get('signal_type', 'Unknown')}

---

## Key Metrics
"""
        
        if analysis_results:
            basic_stats = analysis_results.get('basic_stats', {})
            gradient = analysis_results.get('gradient', {})
            
            slides += f"""
- **Total Variation**: {gradient.get('total_variation', 0):.4f}
- **Signal Range**: {basic_stats.get('signal_range', 0):.4f}
- **Graph Connectivity**: {len(gradient.get('edge_list', []))} edges
"""
        
        slides += """
---

## Spectral Analysis
- Frequency domain decomposition
- Energy distribution analysis
- Filter response evaluation

---

## Conclusions
- Signal processing pipeline executed successfully
- Results exported for further analysis
- Recommendations provided in detailed report

---

*Generated by Bioctree Dashboard*
        """
        
        return slides
    
    def export_complete_session(self, session_name=None):
        """Export complete analysis session"""
        
        if session_name is None:
            session_name = f"bioctree_session_{time.strftime('%Y%m%d_%H%M%S')}"
        
        session_dir = self.export_dir / session_name
        session_dir.mkdir(exist_ok=True)
        
        exported_items = []
        
        # Export current data
        if data_manager.current_graph and data_manager.current_signal is not None:
            
            # Export graph and signal data
            session_data = {
                'graph_metadata': data_manager.metadata,
                'signal': data_manager.current_signal.tolist(),
                'graph_size': data_manager.current_graph.N,
                'export_timestamp': time.strftime('%Y-%m-%d %H:%M:%S')
            }
            
            data_files = self.export_data(session_data, session_dir / "session_data")
            exported_items.extend(data_files)
            
            # Run analysis and export results
            try:
                analysis_results = gsp_processor.run_comprehensive_analysis(
                    data_manager.current_graph, data_manager.current_signal)
                
                # Export analysis results
                analysis_files = self.export_data(analysis_results, session_dir / "analysis_results")
                exported_items.extend(analysis_files)
                
                # Generate and export report
                report = self.generate_analysis_report(
                    analysis_results, data_manager.metadata, data_manager.metadata)
                
                report_path = session_dir / "analysis_report.md"
                with open(report_path, 'w') as f:
                    f.write(report)
                exported_items.append(str(report_path))
                
                print(f"✅ Complete session exported to: {session_dir}")
                
            except Exception as e:
                print(f"⚠️ Error during analysis export: {e}")
        
        else:
            print("❌ No data available for export")
        
        # Update export history
        self.export_history.append({
            'session_name': session_name,
            'timestamp': time.strftime('%Y-%m-%d %H:%M:%S'),
            'items': exported_items,
            'location': str(session_dir)
        })
        
        return exported_items, str(session_dir)

# Initialize export manager
export_manager = ExportManager()

# Interactive export interface
@interact_manual(
    export_session=False,
    export_figures=False,
    generate_report=False,
    session_name='bioctree_analysis'
)
def export_interface(export_session=False, export_figures=False, generate_report=False, session_name='bioctree_analysis'):
    """Interactive export interface"""
    
    if export_session:
        print("💾 Exporting complete session...")
        files, location = export_manager.export_complete_session(session_name)
        print(f"📁 Session exported to: {location}")
        print(f"📄 Files exported: {len(files)}")
    
    if export_figures and data_manager.current_graph is not None:
        print("📊 Exporting visualizations...")
        try:
            # Generate and export 2D visualization
            fig_2d = viz_engine.plot_graph_signal_2d(
                data_manager.current_graph, data_manager.current_signal)
            export_manager.export_figure(fig_2d, f"{session_name}_network_view")
            plt.close(fig_2d)
            
            # Generate and export spectral analysis
            fig_spectral, _ = viz_engine.plot_spectral_analysis(
                data_manager.current_graph, data_manager.current_signal)
            export_manager.export_figure(fig_spectral, f"{session_name}_spectral_analysis")
            plt.close(fig_spectral)
            
        except Exception as e:
            print(f"❌ Error exporting figures: {e}")
    
    if generate_report and data_manager.current_graph is not None:
        print("📝 Generating analysis report...")
        try:
            analysis_results = gsp_processor.run_comprehensive_analysis(
                data_manager.current_graph, data_manager.current_signal)
            
            report = export_manager.generate_analysis_report(
                analysis_results, data_manager.metadata, data_manager.metadata)
            
            print("\\n" + "="*60)
            print("📋 ANALYSIS REPORT PREVIEW")
            print("="*60)
            print(report[:1000] + "..." if len(report) > 1000 else report)
            print("="*60)
            
        except Exception as e:
            print(f"❌ Error generating report: {e}")

print("💾 Export and reporting system ready!")
````

````python
class MasterDashboard:
    """Integrated master dashboard controller"""
    
    def __init__(self):
        self.data_manager = data_manager
        self.gsp_processor = gsp_processor
        self.viz_engine = viz_engine
        self.dev_framework = dev_framework
        self.doc_generator = doc_generator
        self.export_manager = export_manager
        
        self.session_log = []
    
    def run_complete_pipeline(self, graph_type='Stanford Bunny', n_vertices=50, 
                             signal_type='Smooth', noise_level=0.1, 
                             auto_export=False, session_name=None):
        """Run the complete analysis pipeline"""
        
        print("🚀 BIOCTREE DASHBOARD - COMPLETE PIPELINE EXECUTION")
        print("="*70)
        
        # Step 1: Data Generation
        print("\\n📊 Step 1: Generating data...")
        success, message = self.data_manager.generate_graph(graph_type, n_vertices)
        print(f"   {message}")
        
        if success:
            success, message = self.data_manager.generate_signal(signal_type, noise_level)
            print(f"   {message}")
        
        if not success:
            print("❌ Pipeline aborted due to data generation failure")
            return None
        
        # Step 2: Analysis
        print("\\n🔬 Step 2: Running comprehensive analysis...")
        analysis_results = self.gsp_processor.run_comprehensive_analysis(
            self.data_manager.current_graph, self.data_manager.current_signal)
        print("   ✅ Analysis completed")
        
        # Step 3: Visualization
        print("\\n📈 Step 3: Generating visualizations...")
        
        # Create main visualizations
        fig_2d = self.viz_engine.plot_graph_signal_2d(
            self.data_manager.current_graph, self.data_manager.current_signal,
            title=f"{signal_type} Signal on {graph_type}")
        plt.show()
        
        fig_spectral, spectral_data = self.viz_engine.plot_spectral_analysis(
            self.data_manager.current_graph, self.data_manager.current_signal)
        plt.show()
        
        print("   ✅ Visualizations generated")
        
        # Step 4: Pattern Analysis
        print("\\n🔍 Step 4: Pattern analysis and insights...")
        patterns = analysis_results['patterns']['spectral_patterns']
        gradient_info = analysis_results['gradient']
        
        # Generate insights
        insights = []
        if patterns['is_smooth']:
            insights.append("🌊 Signal is SMOOTH - dominated by low-frequency components")
        if patterns['is_oscillatory']:
            insights.append("🌪️ Signal is OSCILLATORY - contains significant high-frequency content")
        if patterns['is_localized']:
            insights.append("🎯 Signal is LOCALIZED - energy concentrated in few frequency bands")
        
        if gradient_info['total_variation'] > 1.5:
            insights.append("📈 HIGH spatial variation detected across graph edges")
        elif gradient_info['total_variation'] < 0.5:
            insights.append("📉 LOW spatial variation - signal is very smooth")
        
        for insight in insights:
            print(f"   {insight}")
        
        if not insights:
            print("   📊 Signal shows balanced frequency characteristics")
        
        # Step 5: Export (if requested)
        if auto_export:
            print("\\n💾 Step 5: Exporting results...")
            if session_name is None:
                session_name = f"{graph_type.lower().replace(' ', '_')}_{signal_type.lower()}_{time.strftime('%H%M%S')}"
            
            files, location = self.export_manager.export_complete_session(session_name)
            print(f"   📁 Exported to: {location}")
        
        # Step 6: Summary
        print("\\n📋 Step 6: Executive Summary")
        print("-" * 50)
        
        basic_stats = analysis_results['basic_stats']
        print(f"🔹 Graph: {graph_type} ({basic_stats['graph_size']} vertices)")
        print(f"🔹 Signal: {signal_type} (noise: {noise_level})")
        print(f"🔹 Total Variation: {gradient_info['total_variation']:.4f}")
        print(f"🔹 Spectral Energy Distribution:")
        print(f"   • Low frequencies: {patterns['low_freq_energy']:.1%}")
        print(f"   • High frequencies: {patterns['high_freq_energy']:.1%}")
        
        # Log session
        session_entry = {
            'timestamp': time.strftime('%Y-%m-%d %H:%M:%S'),
            'graph_type': graph_type,
            'signal_type': signal_type,
            'results': analysis_results,
            'insights': insights
        }
        self.session_log.append(session_entry)
        
        print("\\n✅ PIPELINE COMPLETED SUCCESSFULLY!")
        print("="*70)
        
        return analysis_results
    
    def create_master_control_panel(self):
        """Create the master control panel with all features"""
        
        # Main control widgets
        graph_controls = widgets.VBox([
            widgets.HTML("<h3>🌐 Graph Generation</h3>"),
            widgets.Dropdown(
                options=['Stanford Bunny', 'Random Graph', 'Grid Graph', 'Ring Graph'],
                value='Stanford Bunny',
                description='Type:'
            ),
            widgets.IntSlider(
                value=50, min=10, max=200, step=10,
                description='Vertices:'
            )
        ])
        
        signal_controls = widgets.VBox([
            widgets.HTML("<h3>📡 Signal Generation</h3>"),
            widgets.Dropdown(
                options=['Smooth', 'Oscillatory', 'Localized', 'Random'],
                value='Smooth',
                description='Type:'
            ),
            widgets.FloatSlider(
                value=0.1, min=0.0, max=0.5, step=0.05,
                description='Noise:'
            )
        ])
        
        action_controls = widgets.VBox([
            widgets.HTML("<h3>🎯 Actions</h3>"),
            widgets.Button(description="🚀 Run Pipeline", button_style='primary'),
            widgets.Button(description="📊 Quick Analysis", button_style='info'),
            widgets.Button(description="💾 Export All", button_style='warning'),
            widgets.Button(description="🧪 Run Tests", button_style='success')
        ])
        
        # Status and output
        status_output = widgets.Output()
        
        # Complete dashboard
        master_panel = widgets.VBox([
            widgets.HTML("<h1 style='text-align:center'>🧠 BIOCTREE MASTER DASHBOARD</h1>"),
            widgets.HTML("<p style='text-align:center'>Central Control for Graph Signal Processing</p>"),
            widgets.HTML("<hr>"),
            widgets.HBox([graph_controls, signal_controls, action_controls]),
            widgets.HTML("<hr>"),
            status_output
        ])
        
        return master_panel
    
    def get_session_summary(self):
        """Get summary of all sessions"""
        
        if not self.session_log:
            return "No sessions recorded yet."
        
        summary = f"""
        📊 **Dashboard Session Summary**
        
        Total Sessions: {len(self.session_log)}
        Last Session: {self.session_log[-1]['timestamp']}
        
        **Recent Activity:**
        """
        
        for i, session in enumerate(self.session_log[-5:]):  # Last 5 sessions
            summary += f"""
        {i+1}. {session['timestamp']} - {session['graph_type']} + {session['signal_type']}
           Insights: {len(session['insights'])} detected
            """
        
        return summary

# Initialize master dashboard
master_dashboard = MasterDashboard()

# Quick action buttons
def run_demo_pipeline():
    """Run a quick demo of the complete pipeline"""
    return master_dashboard.run_complete_pipeline(
        graph_type='Stanford Bunny',
        signal_type='Smooth', 
        noise_level=0.1,
        auto_export=False
    )

def run_advanced_demo():
    """Run advanced demo with different parameters"""
    return master_dashboard.run_complete_pipeline(
        graph_type='Random Graph',
        signal_type='Oscillatory',
        noise_level=0.2,
        auto_export=True,
        session_name='advanced_demo'
    )

# Display master control interface
@interact_manual(
    demo_basic=False,
    demo_advanced=False,
    show_session_log=False,
    run_diagnostics=False
)
def master_control_interface(demo_basic=False, demo_advanced=False, 
                           show_session_log=False, run_diagnostics=False):
    """Master control interface for the dashboard"""
    
    if demo_basic:
        print("🎮 Running basic demo pipeline...")
        run_demo_pipeline()
    
    if demo_advanced:
        print("🎮 Running advanced demo pipeline...")
        run_advanced_demo()
    
    if show_session_log:
        summary = master_dashboard.get_session_summary()
        display(Markdown(summary))
    
    if run_diagnostics:
        print("🔧 Running system diagnostics...")
        
        # Test all components
        print("   📊 Data Manager: ", "✅ OK" if data_manager else "❌ Error")
        print("   🔬 GSP Processor: ", "✅ OK" if gsp_processor else "❌ Error")
        print("   📈 Visualization: ", "✅ OK" if viz_engine else "❌ Error")
        print("   🧪 Dev Framework: ", "✅ OK" if dev_framework else "❌ Error")
        print("   📚 Documentation: ", "✅ OK" if doc_generator else "❌ Error")
        print("   💾 Export Manager: ", "✅ OK" if export_manager else "❌ Error")
        
        # Test basic functionality
        try:
            test_graph = stanford_bunny()
            test_signal = np.random.randn(test_graph.N)
            print(f"   🧪 Basic GSP Test: ✅ OK (Graph: {test_graph.N} vertices)")
        except Exception as e:
            print(f"   🧪 Basic GSP Test: ❌ Error - {e}")
        
        print("\\n🎯 Dashboard is ready for operation!")

print("\\n" + "="*70)
print("🎉 BIOCTREE INTERACTIVE DASHBOARD FULLY LOADED!")
print("="*70)
print("📋 Available Components:")
print("   🌐 Data Management     📊 Visualization Engine")
print("   🔬 Signal Processing   🧪 Development Framework") 
print("   📚 Documentation       💾 Export & Reporting")
print("\\n🚀 Use the master control interface above to get started!")
print("💡 Or run individual components using their respective interfaces.")
print("="*70)
````
